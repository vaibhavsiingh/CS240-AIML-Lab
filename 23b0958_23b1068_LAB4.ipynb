Function to automate data generation

```python
def generate_perm(n):
    all_perm = []
    for mask in range(1<<n):
        sequence = [(mask >> i) & 1 for i in range(n-1,-1,-1)]
        all_perm.append(sequence)
    return all_perm
    
```

```python
def is_palindrome(seq):
    return seq == seq[::-1]

def check_palindrome(X):
    return [is_palindrome(seq) for seq in X]

X_palindrome_4 = generate_perm(4)
Y_palindrome_4 = check_palindrome(X_palindrome_4)
Y_palindrome_4 = np.array(Y_palindrome_4).reshape(-1,1).astype(int)
X_palindrome_4 = np.array(X_palindrome_4)
```

```python
def is_majority(seq):
    n_ones = 0
    n_zeros = 0
    for bit in seq:
        if bit == 1:
            n_ones+=1
        else:
            n_zeros+=1
    #print(seq, n_zeros, n_ones)
    if n_ones > n_zeros:
        return 1
    else:
        return 0

def majority(X):
    return [is_majority(seq) for seq in X]

X_majority_4 = generate_perm(4)
Y_majority_4 = majority(X_majority_4)
Y_majority_4 = np.array(Y_majority_4).reshape(-1,1).astype(int)
X_majority_4 = np.array(X_majority_4)
```

```python
def is_parity(seq):
    n_ones = 0
    for bit in seq:
        if bit == 1:
            n_ones+=1
    #print(seq, n_zeros, n_ones)
    if n_ones %2==0:
        return 1
    else:
        return 0

def parity(X):
    return [is_parity(seq) for seq in X]

X_parity_4 = generate_perm(4)
Y_parity_4 = parity(X_parity_4)
Y_parity_4 = np.array(Y_parity_4).reshape(-1,1).astype(int)
X_parity_4 = np.array(X_parity_4)
X_parity_4, Y_parity_4
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
def sigmoid_derivative(x):
    return x*(1-x) 
```

# Class for feed forward neural network

```python
class FeedForwardNN:
    def __init__(self,input_size):
        self.weights = np.random.rand(input_size,1)
        self.bias = np.random.rand(1)
        
    def forward(self, X):
        z = np.dot(X,self.weights)+self.bias
        return sigmoid(z)

    def mse_loss(self,y, y_pred):
        return np.sum((y-y_pred)**2)

    def backward(self,X,Y,Y_pred,learning_rate):
        dloss_dz = 2*(Y_pred-Y)*sigmoid_derivative(Y_pred)
        dloss_dw = np.dot(X.T,dloss_dz)/X.shape[0]
        dloss_db = np.sum(dloss_dz)/X.shape[0]

        self.weights -= learning_rate*dloss_dw
        self.bias -= learning_rate *dloss_db

    def predict(self,X):
        return [1 if self.forward(seq) > 0.5 else 0 for seq in X]
   
```

```python
 def train(X,Y,iterations=1000,learning_rate=0.1):
        input_size = X.shape[1]
        ffnn = FeedForwardNN(input_size)
        mse_loss = []

        for i in range(iterations):
            Y_pred = ffnn.forward(X)
            loss = ffnn.mse_loss(Y,Y_pred)
            mse_loss.append(loss)
            ffnn.backward(X,Y,Y_pred,learning_rate)

        final_pred = ffnn.predict(X)
      #  print(final_pred,mse_loss)
        return final_pred, mse_loss
        
```
